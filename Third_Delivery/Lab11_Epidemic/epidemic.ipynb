# Epidemic Process

The objective of Lab L11 is to formulate and simulate a simple multi-group SIR-type model that incorporates population inhomogeneity, including differences in risk of exposure, fatality rates, and spatial heterogeneity.

The model must also capture the effects of public-health policies aimed at detecting and isolating infected but asymptomatic individuals. In addition, the model should account for limited natural immunity.

You are required to define all model variables and parameters explicitly and to derive the corresponding mean-field equations. After establishing the model, implement it computationally and analyze how selected parameters influence the epidemic dynamics.

Finally, assume that vaccines become available and that they can be produced and administered at a fixed maximum rate. Extend the model to incorporate vaccination and to evaluate its impact on the epidemic trajectory. Develop and compare several vaccination strategies, where each strategy specifies the priority order in which population groups receive the vaccine. Assess the outcomes of these strategies and examine how they interact with non-pharmaceutical interventions (e.g., mobility restrictions) that may be applied during the vaccination campaign to mitigate epidemic spread.

## Part 1: Model Definition & Derivation
*Objective: Formulate the mathematical framework for a heterogeneous population with waning immunity and isolation policies.*

### 1.1 State Variables (Compartmental Structure)
Define the compartments for a specific group $k$ (where $k$ represents a spatial region, age group, or risk category). To meet the requirements (asymptomatic isolation, limited immunity, fatality), a suggested structure is **SEIQRS** or **SAI-Q-RS**:
* $S_k$: Susceptible.
* $E_k$ (or $A_k$): Exposed/Asymptomatic (Infected but not yet symptomatic/detected).
* $I_k$: Symptomatic Infected.
* $Q_k$: Quarantined/Isolated (Detected asymptomatic or symptomatic individuals).
* $R_k$: Recovered (Temporary immunity).
* $D_k$: Deceased (Cumulative count).

### 1.2 Parameters & Heterogeneity
Define parameters explicitly, noting which are group-dependent:
* **Transmission Matrix ($\beta_{kl}$):** Rate at which group $l$ infects group $k$ (captures spatial heterogeneity and contact patterns).
* **Disease Progression:** Latency rates ($\sigma$), recovery rates ($\gamma$), and fatality probabilities ($\delta_k$ - group specific risk).
* **Immunity Loss:** Waning immunity rate $\xi$ (flow from $R \to S$).
* **Public Health Policy:** Detection/Isolation rate $\theta_k$ (rate of moving from $A/I \to Q$).

### 1.3 Mean-Field Equations (Derivation)
Write the system of Ordinary Differential Equations (ODEs) for group $k$.
* *Key feature to demonstrate:* The force of infection $\lambda_k$ must sum contributions from all other groups $l$:
    $$\lambda_k(t) = S_k \sum_{l=1}^{C} \beta_{kl} (I_l + \epsilon A_l)$$
* *Key feature to demonstrate:* The flow from Recovered back to Susceptible ($R_k \to S_k$) to model limited natural immunity.

## Part 3: Vaccination Strategies (The Extension)
*Objective: Model resource-constrained vaccination and optimize allocation.*

### 3.1 Extended Model with Vaccination
* Introduce a vaccination flux: $v_k(t)$ representing individuals moving from $S_k \to R_k$ (or a new $V_k$ compartment).
* **Constraint:** The total vaccination rate across all groups is bounded by production capacity $V_{max}$:
    $$\sum_{k=1}^{C} v_k(t) \leq V_{max}$$

### 3.2 Defining Strategies
Define priority queues for distributing the limited $V_{max}$:
1.  **Strategy A (Mortality Minimization):** Prioritize groups with the highest fatality rate $\delta_k$ (e.g., elderly/vulnerable).
2.  **Strategy B (Transmission Reduction):** Prioritize groups with the highest contact rates or centrality in the interaction matrix (e.g., "super-spreaders").
3.  **Strategy C (Prevalence Based):** Prioritize regions with the fastest growing infection rates.

### 3.3 Strategy Assessment & NPI Interaction
* **Comparative Plots:** Plot Total Deaths vs. Time for Strategy A, B, and C.
* **Synergy with NPIs:** Run the best vaccination strategy alongside a "Lockdown" scenario (scaling down $\beta_{kl}$ by factor $\alpha < 1$). Determine if NPIs are still necessary during the vaccine rollout to prevent system collapse.
